#### 1.  Square Star Pattern


```python
rows=5

for i in range(rows):
    for j in range(rows):
        print("*",end="")
    print()
```

#### 2. Pyramid Pattern

```python
rows=5
for i in range(1,rows+1):
    for j in range(rows-i):
        print("",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()
```

#### 3. Diamond Pattern

```python
rows=5
for i in range(1,rows +1):
    for j in range(rows-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()
for i in range(rows -1,0,-1):
    for j in range(rows -i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()
```

#### 4. Number Pyramid

```python
rows=5

for i in range(1,rows +1):
    for j in range(rows-i):
        print(" ",end=" ")
    for k in range(1,i+1):
        print(k,end=" ")
    print()
```

#### 5. Butterfly Pattern

```python
for i in range(1,rows+1):
    print("*"*i,end="")
    print(" " *(2*(rows-i)),end="")
    print("*"*i)
for i in range(rows-1,0,-1):
    print("*"*i,end="")
    print(" " *(2*(rows-i)),end="")
    print("*"*i)
```

#### 6. Small Alphabet Triangle

```python
rows=5
ch=ord('a')
for i in range(1,rows +1):
    for j in range(i):
        print(chr(ch),end=" ")
        ch+=1
    print()
```

#### 7. Same small Letter Pattern

```python
rows=5
for i in range(rows):
    ch=chr(97+i)
    for j in range(i+1):
        print(ch,end=" ")
    print()
```

#### 8. Alphabet Square Pattern

```python
rows=5
for i in range(rows):
    for j in range(rows):
        print(chr(97+j),end=" ")
    print()
```

#### 9. Reverse Small Alphabet Pattern

```python
rows=5
for i in range(rows):
    for j in range(rows - i):
        print(chr(97+j),end=" ")
    print()
```
